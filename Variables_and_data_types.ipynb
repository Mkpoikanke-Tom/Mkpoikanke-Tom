## Re-assigning Variables:
 A variable can change it value during the excution of a program, for example the value of the variable called name was mkpoikanke then the same variable is is assigned a new value which replaces the old one as Elisha, if we print the variable name what come up as the output becomes the new value. This means the new value of the variable replaces the old one.So this is to say that a variable reassignment overwrites the previous value stored in the variable if not printed because the variable is assigned to another value, that is python displays the most recent value because the previous value has been overwritten.

```python
##Write a python program to showcase this

name = "Mkpoikanke"
name = "Elisha"
print(name)
```

```python
##Another example of a variable re-assignment is the code below

name = "Tom"
age = 25
print(name)
age = 45
name ="Daniel"
print(age)
print(name)
```

## Dynamic Typing:
In python variables can change not only its values but data types.This is an important concept to realize in python because other programming languages like java an c++ does not do same because the are statistically typed. So dynamic typing is a feature in python where a variable can store different type of data at different times during the excution of a program, this is to say that python automatically detects the data type of the value assigned to the variable, so we need not to declare the variable type beforehand like in other programming languages such as java and c++. For instance take a variable as a locker, then we put in a book, money and probably a football inside , then lets call the book a string, the money an integer and then the football a boolean value showing the different data types in the locker which is the variable, in python we will see that each time we excute the program python automatically detects the data type and prints it according to what its detects as a value, but in java when the variable is defined beforehand as an integer what happens is that if the value stores in the variable is a string, during excution java returns error.
## NOTE:
 In python the value decides the type of variable at that moment, but in java for instance the variable decides what kind of data can be stored.



```python
##Under dynamic typing, let call the variable item

item = "Book"
print(item)

item = 500
print(item)

item = True
print(item)
```

```python
##Combinning variable reassignment with dynamic typing; this shows different data type in the same variable and returns the new value its detects.

age = 26
age = "Twenty one"
age = 24.5
print(age)
```

##Type Conversion;
 Type conversion is the process of changing value from one data type to another. For example
1. A string can become an integer.
2. An integer can become a float.
3. A float can become a string.








```python
##Four main types of conversion
# converting from a string to an integer
number = int("30")
print(number)

#converting from a string to a floating-point number
figure = float("150")
print(figure)

# converting from an integer to a string
age = 21
text = str(age)
print(text)

# converting from an integer to a boolean value
is_age = 35
convert = bool(is_age)
print(convert)
print(bool(1))
print(bool(0))

## Type conversion does not change the variable unless you assign the result back
# For instance
age = "35"
print(int(age))

# Note that if the new variable is assigned to the previous variable, the data type changes to the data type it is being converted to for instance

age = 26
converted = str(age)
print(converted)
print(type(converted))
```

```python
# why type-conversion is important
age = "30"
added_years = 5
total_years = int(age) + added_years
print(total_years)
```

```python
number =  "12"
add_number = 0.5
total = float(number) + add_number
print(total)

days = 100
text = "days"
full_text = str(days) + " " + text
print(full_text)
```
