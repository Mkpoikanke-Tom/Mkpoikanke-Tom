```
## A python program that asks the user to enter their account account_balance and the withdrawal_amount the want top redraw, the program does the following
## If withdrawal withdrawal_amount is less than or equal to the account_balance, print '"Withdrawal Successful.,Remaining account_balance:$..... " otherwise print insufficient fund

account_balance_withdrawal_amount = int(input("Enter account account_balance:"))
withdrawal_withdrawal_amount = int(input("Enter withdrawal withdrawal_amount:"))
if withdrawal_withdrawal_amount <= account_balance_withdrawal_amount:
  account_balance = account_balance_withdrawal_amount - withdrawal_withdrawal_amount
  print(f"withdrawal successful, Remaining account_balance is: ${account_balance}")
else:
  print("Insufficient Fund")
```

```
##Simple banking system: ask the user to enter acoount account_balance and withdrawal withdrawal_amount then write an if condition

account_balance = int(input("Enter account account_balance:"))
withdrawal_amount = int(input("Enter withdrawal withdrawal_amount:"))
naira = "\u20A6"
if withdrawal_amount <= 0:
  print("Invalid withdrawal withdrawal_amount")
elif withdrawal_amount > account_balance:
  print("Insufficient funds")
else:
  remaining_account_balance = account_balance - withdrawal_amount
  print(f"Withdrawal successful,Remaining account_balance:{naira}{remaining_account_balance}")
  if remaining_account_balance < 1000:
    print("Warning:Low account_balance")
```
