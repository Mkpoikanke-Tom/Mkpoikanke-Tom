##** Conditional Statement**
 A conditional statement is used to make decision either when a conditions true that is when its just an "if" statement or to execute one block  if the conditon is true or another block if the condition is false
 ## **Different conditional statements**


*   if statement; this is a decision making statement that executes the block of code only when the specified condition is true.
*   if...else statement;this decision statement  excutes one block of code if the condition is True or another block of code if the conditon is false.
*   if...elif...else statement; this statement is used when there are multiple conditions to check.Python tests each condition from button to top and excutes only the first condition that is **true**, if none are true its executes the **else** block.  
*   Nested-if-statement;a nested if statement is an if statement inside another if statement. This is when python checks one condition first,if that condition is True it then checks the other conditon inside it.
Note that every condition end with a colon(:).



```python
# How if statements work
# if entells "i want to make a decision" after which a comparison operator is being used
# to spell the condition to which the decision is being made.For instance

age = 20
if age >= 18:
  print("You are an adult")        ##python executes the the print statement because the condition here is true
```

## **Logical Operators**
Logical operators are operators used to modify or combine two or more conditions.They return a Boolean value(True or False).

*   AND operator: the AND operator returns True only when all he conditions being tested are True. this is to say that if any one condition is False, the entire expression becomes false.
*   OR operator: the OR operator returns True if atleast one of the condition is True. it returns false only when all the conditions are false.
*   NOT Operator: the NOT operator reverses the Boolean value of a condition. It changes True to False and False to True.

## Note that:
Logical operators are often used in conditional statements(if, if...else, if...elif...else and Nested if statements) to evaluate multiple conditions or reverse a condition thereby allowing python to make more complex decisions.



```python
## Practical examples
# for AND operator

age = 20
has_id = True

if age >= 18 and has_id:     #Python executes the if block since the condition are all true
  print("Entry Allowed")



# OR operator
age = 15
has_permission = True

if age >= 18 or has_permission:    #Python executes the if block since one of the condition is met
  print("Entry Allowed")



# NOT operator
logged_in = False
if not logged_in:              #python executes the if block because the NOT operator reverses the if condition from False to True
  print("Please log in")
```

```python
## Examples of for loop
for i in range(1,6):  #python evaluate this as a range between 1 to 5, this is to say that and excludes 6 which is the stopping point
  print(i)            # note that range(start,stop)
```

```python
# Nested for loop:
# a nested for loop is a for loop inside another for loop. the inner loop completes all of its iterations for every single iteration of the outer loop.
# the outer for loop controls the number of rows.
# The inner for loop control how many values are printed in each row.
# For example

for i in range (1,5):     #the outer loop starts with i = 1 and goes up to 4
                          # For every value of i, the inner loop start from j = 1 and goes up to 4 again

  for j in range (1,5):   # The inner loop (j) runs completely before the outer loop moves to the next value of i(next row)
                          # although j changes from 1 to 4, we are printing i, so the same value of i is printed 4 times on the same line.
    print(i, end = "")    # after the inner loop finishes move to the next line(that is 1 to 4 at each i).
                          # so the next row start on a new line
  print()


# Trace:
#
# i = 1
#     j = 1 -> print(1)
#     j = 2 -> print(1)
#     j = 3 -> print(1)
#     j = 4 -> print(1)
# Output: 1111
#
# i = 2
#     j = 1 -> print(2)
#     j = 2 -> print(2)
#     j = 3 -> print(2)
#     j = 4 -> print(2)
# Output: 2222
#
# i = 3
#     j = 1 -> print(3)
#     j = 2 -> print(3)
#     j = 3 -> print(3)
#     j = 4 -> print(3)
# Output: 3333
#
# i = 4
#     j = 1 -> print(4)
#     j = 2 -> print(4)
#     j = 3 -> print(4)
#     j = 4 -> print(4)
# Output: 4444
```

```python
## a right angled triangle pattern program

for i in range(1,5):   # The outer loop controls the number of row.

  for j in range(i):   # The inner loop depends in the number of the i
                       # so as i increases, the inner loop runs more times,
                       # so each new row prints one more star than the previous one.

    print("*",end="")

  print()               # Print() moves to the next line after each row is completed
```

```python
# An inverted-right angled triangle
for i in range(5, 0, -1):
  for j in range(i):
    print("*",end="")
  print()
```
