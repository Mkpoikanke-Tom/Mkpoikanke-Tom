## Conditional Statement
 A conditional statement is used to make decision either when a conditions true that is when its just an "if" statement or to execute one block  if the conditon is true or another block if the condition is false
 ## Different conditional statements includes


*   if statement; this is a decision making statement that executes the block of code only when the specified condition is true.
*   if...else statement;this decision statement  excutes one block of code if the condition is True or another block of code if the conditon is false.
*   if...elif...else statement; this statement is used when there are multiple conditions to check.Python tests each condition from button to top and excutes only the first condition that is **true**, if none are true its executes the **else** block.  
*   Nested if statement
Note that every condition end with a colon(:).



```python
# How if statements work
# if entells "i want to make a decision" after which a comparison operator is being used
# to spell the condition to which the decision is being made.For instance

age = 20
if age >= 18:
  print("You are an adult")        ##python executes the the print statement because the condition here is true
```

```python

## what if we had the same variable age with false condition, lets see what python returns

age = 20
if age < 18:
  print("You are an adult")     # We see that python skips the indented line and nothing is printed,
                                # this is where the else condition comes in as returning another block if the condition specified is false
```

```python
##Another example of an if condition
marks = 45
if marks >= 50:                            #Python prints the second block because the if condition is false, which thereby passes the indented line
  print("Pass")                            # and prints the second statement. Note that the line printed is outside the if statement.
print("End of program")                    # so it returns whether the condition is true or false




##Important rule
# condition = []
# if condition:
      #Runs only if the condition is true

#This line is outside the if statement above, thus it runs whether the condion is true or false

```

```python
## If...else statement

age = 16

if age >= 18:                      #Think of this like, if this is true, do this, otherwise do that.
  print("You can vote.")           #Recall earlier that if conditon returns its line when the condition is true,
else:                              #but what happened here is that since the if condition false,python returns another block since the condition is false.
  print("You cannot vote yet.")


# Python syntax
# if condition:
#   prints statement if true
#  else:
#    prints statement if false


```

```python
## if..elif...else statement
# grades for students according to their scores

scores = 65
if scores >= 70:        # False
  print("A")
elif scores >= 60:      # True
  print("B")
elif scores >= 50:      # Not checked, this is because when python finds the true conditon its doesn't evaluate the rest
  print("C")
elif scores >= 40:      # Not checked, python does not evaluate
  print("D")
else:
  print("F")            # Notchecked, python does not evaluate
```

```python
##The use of input to imput the scores of student manually to generate their grade in an if..elif..else

scores = int(input('"Enter your scores:'))
if scores >= 70:
  print("Your grade is A.")
elif scores >= 60:
  print("Your grade is B.")
elif scores >= 50:
  print("Your grade is C.")
elif scores >= 40:
  print("your grade is D.")
else:
  print("your grade is F.")
```
