```python
## Modifying a list through indexing
numbers = [10,20,30,50,70]
numbers.append(40)
numbers[3] =40
numbers[4] =50
numbers[5] =70
print(numbers)
```

```python
##Finding the largest number in the list
numbers = [18,7,50,22,31]
largest = numbers[0]
for number in numbers:
    if number > largest:
        largest = number
print(largest)
```

```python
##Finding the smallest number in the list

numbers =[18,2,50,22,31]
smallest = numbers[0]
for number in numbers:
  if number < smallest:
    smallest = number
print(smallest)
```

```python
## Adding the total number of items on the list
numbers =[18,2,50,22,31]
total = 0
for number in numbers:
    total += number
print(total)
```

```python
## Computing the average of an item on the  list
numbers =[18,2,50,22,31]
total = 0
for number in numbers:
    total += number
    average = total / len(numbers)
print(average)
```

```python
## counting numbers greater than a given number in a list
numbers = [4,12,7,18,9,15,20]
count = 0

for number in numbers:
  if number > 10:
    count += 1
print(count)

```

```python
## Counting odd numbers in the list
numbers = [6, 10, 12, 15, 17, 18, 22, 24]
count = 0
for number in numbers:
  if number % 3 == 0:
    count += 1
print(count)
```

```python
##Using the "AND" operator to locate even number greater than 10 in the list
numbers = [4, 12, 7, 18, 9, 15, 20, 22]
count = 0
for number in numbers:
  if number % 2 == 0 and number > 10:
    count += 1
print(count)
```

```python
## Using an 'OR' operator in the list
numbers = [2, 6, 18, 3, 9, 20, 5, 1]
count = 0
for number in numbers:
  if number < 5 or number > 15:
    count += 1
print(count)
```

```python
##Combining this three operators "NOT", "AND" operator
numbers = [3, 6, 11, 14, 17, 20, 25, 30]
count = 0
for number in numbers:
  if number % 2 != 0 and number > 10:
    count += 1
print(count)
```

```python
## A program to count the number of even numbers and numbers that are divisible by three in the list

numbers = [6, 8, 12, 15, 18, 20, 24, 27]
count = 0
for number in numbers:
  if number % 2 == 0 and number % 3 == 0:
    count += 1
print(count)
```

```python
##A program that counts odd, divisible by 5 , greater than 20 in a list

numbers = [5, 10, 15, 21, 25, 30, 35, 41, 45]
count = 0
for number in numbers:
  if number % 2 != 0 and number % 5 == 0 and number > 20:
    count += 1
print(count)
```
