##** Function**
A function is a reusable block of code that performs a specific task. Instead of writing the same code over and over again,you write it inside a function and call it whenever you need it.
## **Uses of functions**
*   Functions are used to reuse code instead of writing it repeatedly.
*   Function are used to make programs easier to read.
*   There are used to break a large problem into smaller tasks.
*   There are used to make debugging easier.
*   Functions are also used to make programs easier to maintain.
  







```python
## Function example

def greet():             # "def" is a python keyword which tells python i want to create a function,
  print("Hello Tom")     # "greet" this is the name of the function, just like variable have names
                         # () this is called parentheses.
                         # : the colon tell python that everything indented below belongs to the function.
                         # Note that the code there is going to print nothing because the function has not been called.
```

```python
# The same code when the funcyion is been called
def greet():
  print("Hello Tom")      # This (Hello Tom) is what python store inside the function.

greet()                   # We notice here that python execute everything inside the function because it has been called.
                          # Note that defining a function does not execute it. A function is only executed when it is called using its name followed by parentheses.
```

```python
## Execises to illustrate the callability of functions

def login():
  print("Login Successful")
def dashboard():
  print("Dashboard opened")
def setting():
  print("Setting Opened")
def logout():
  print("Logged Out")

print("Start")        # Python print start because the fuction create are not called.

login()               # This are the called functions
dashboard()
setting()
logout()

print("End")         # pythoin execute the last statement after executing the values that are contained in the function.
```

```python
## RETURN: The return send a value back from a function to the place the function was called.
# print() shows the result on the screen
# return() gives the result back so the program can use it.
# for insance


def add():
  return 5 + 4               # Python calculate 5 + 4 = 9.
                             # The keyword 'return' sends the value 9 back to wherever the function is called.

answer = add()               #Python calls the function add(),
                             # the fuction returns 9.
                             # The returned value is then stored inside a variable called answer

print(answer)                # Python prints the value stored in the variable 'answer'.
```

## **Global and local variable in a function**
A global variable is a variable that is created outside any defined function. It is called a global variable because its can be accessed anywhere in the program unless a local variable with the same name temporary hides it.

A local variable is a variable created inside a function. It only exist while the function is executing. Once the function finishes, python deletes the local variable.

```python
# Examples on how this variable works

# Global variable
name = "Tom"

# Function definition
def greet():

    # Local variable
    name = "John"

    # Python prints the local variable because it looks
    # inside the function first.
    print("Inside Function:", name)

# Call the function
greet()

# The local variable has been deleted.
# Python now prints the global variable.
print("Outside Function:", name)
```

```python
# when Function uses a global variable

# Global variable
city = "Uyo"

# Function definition
def display():
                                     # No local variable named country exists. print("City:", city)
   print("Your city:", city)        # Python therefore uses the global variable.print("City:", city)

# Call the function
display()

# Python still prints the same global variable.
print(city)
```

```python
## Two functions
# Global variable
x = 100

def first():

    # Local variable
    x = 50

    # Prints the local variable
    print("First Function Variable:", x)

def second():

    # No local variable exists here.
    # Python uses the global variable because there is no local variable.
    print("Second Function Variable:", x)

# Call both functions
first()
second()

# Print the global variable
print("Global Variable:", x)
```

## Default Parameters
A default parameter are parameters that already has a value. This is to say that if you don't pass an argument, python uses the default value.

```python
## Example
def greet(name = 'Tom'):     #Function definition with a default parameter.

  print("Hello", name )

greet()        #since no argument is passed in the function called, python returns the default value
```

```python
## when an argument is passed
def greet(name = 'Tom'):     #Function definition with a default parameter.

  print("Hello", name )

greet("Grace")        #python prints the value stored in the name i.e the new argument passed when the function is called.
```
