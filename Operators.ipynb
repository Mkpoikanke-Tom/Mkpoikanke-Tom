## **Operators**
Operators are symbols that tell python to perform an operation for example 3 + 4 which equal 7, the operator there is  "+",
## **Types of Operators**



*   Arithmetic Operators
*   Assignment Operatos
*   Comparison Operators
*   Logical Operators
*   Membership Operators
*   Identity Operator

   Arithmetic Operator; this are operators that are used for calculations.The include addition(+), subtraction(-), Multiplication(*), division(/), floor division(//), modulus(%) and exponent(**).






```python
## each operator, how there are used and the output it returns
#Addition(+)

a = 5
b = 7
print(a + b)

#Subtraction(-)
a = 2
b = 3
print(b - a)

#Multiplication(*)
a = 4
b = 6
print(a * b)

#division(/)
a = 9
b = 3
result = a/b
print(result)      ##Notice that the division operator returns a floating-point number not an integer.

#floor division(//)
a = 17
b = 3
result = a//b
print(result)     ##Notice that its returns the whole number quotient after discarding the decimal part.

#Modulus(%):
a = 17
b = 3
result = a % b   ##Notice that this returns the remainder of the arithmetic calculation.
print(result)

#Exponent(**)
a = 3
b = 3
result = a**b
print(result)

score = 40
score += 10
score //= 5
score %= 3
score **= 2
print(score)
```

```python
##Operator Precedence
# Python does not calculate expression from left to right but it rather follows a rule called "BODMAS", just like in the normal calculation.
# BODMAS signifies; Bracket(), Order/Exponent**,Division/,//,%, Multiplication *, Addition +, Subtraction -.
# examples

print(5 + 3 * 2)   #Python evaluates multiplication before addition because * has higher precedence than +.
                   # 3 * 2 = 6
                   # 5 + 6 = 11

print((5 + 3) * 2)   #Here bracket is evaluated before multiplication.
                     # (5 + 3) = 8
                     # 8 * 2 = 16

print(2 ** 3 * 4)  #we also observe that the calculation of the expression starts from the exponent before the multiplication part.
                   # 2^3 = 8
                   # 8 * 4 = 32

print(20 / 5 + 2)   #We observe that the calulation starts from that division part which equals a floating-point number and ends with the addition part
                    # 20 / 5 = 4.0
                    # 4.0 + 2 = 6.0


print(30 - 18 // 3 * 2 + 5)  #What it does is
                             # 18//3 = 6
                             # 6 * 2 = 12
                             # 30 - 12 = 18
                             # 18 + 5 = 23
```

## Assignment Operators**
 Assignmment operators are special operators used to assign a value to a variable or update the current value of a variable by performing an operation and storing the result back in the same variable.



*   Assignment(=).
*   Additional assignment (+=).
*   Subtraction assignment (-=).
*   Multiplication assignment (*=).
*   Division assignment(/=).
*   Floor division assignment (//=).
*   Modulus assignment (%=).
*   exponent assignment (**=).




   







```python
##Assigment operator and how there are used in python

#Assignment(=)
age = 20                                #this assigns a variable to a value.
print(age)



#Additional assignment(+=)
number = 15
number += 5                               #This means number = number + 5.
print(number)                             #Thus python evaluates it as 15 + 5.



#Subtraction assignment(-=)
count = 18
count -= 6                                #Python evaluates it as count = count - 6
print(count)                              # which equals 12


#Division assignment(/=)                 #Python evaluates it as age = age / 4
age /= 4                                 # 20 / 4 which equals 5.0 (float)
print(age)


#Multiplication assignment(*=)           #Python evaluates it as salary = 50000 * 10 and add the naira sign from the unicode
salary = 50000
salary *= 10
print(f"\u20A6{salary}")


#Floor dvision assignment(//=)          #python evaluates it as count = 25 // 4, then it returns the whole number and discards the decimal part
count = 25
count //= 4
print(count)


#

```

 ## **Comparison operators**
 Comparison operators are operators used to compare two values.The result of the comparison is always either True of False.
## Comparison operators include
*   less than(<); for instance 4 < 6 returns "True" or vice versa which returns "False.

*   greater than (>); for instance 2 > 3 returns "False".
*   equals to (==); for instance 5 == 5 returns "True".Note that an assignment(=) operator is different from comparison operator (==).
*   not equal to (!=); for instance 3 != 4 returns "True".
*   less than or equal to (<=); for instance 5 <= 5 returns "True".
*   greater than or equal to (>=); for instance 6 >= 5 returns "True".




```python
## Comparison operator example
##A comparison operator compares two values to return a boolean value(True or False)

c = 3 > 4
print(c)
b = 5 <= 4
print(b)
h = 4 <= 5
print(h)
p = 5 != 6
print(p)

age = 18
print(age == 18)
print(age!= 18)
print(age > 18)
print(age >= 18)
```
